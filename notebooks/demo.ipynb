# Quantum Image Edge Detection (QSED) Interactive Demo Notebook

This notebook demonstrates the complete 6-stage **Quantum Image Edge Detection based on Eight-Direction Sobel Operator for NEQR** proposed by Liu & Wang (2023).

```python
import sys
import os
import numpy as np
import matplotlib.pyplot as plt

# Add project root to sys.path
sys.path.insert(0, os.path.abspath('..'))

from src.classical.image_loader import generate_synthetic_image, save_image
from src.quantum.neqr import NEQRImage
from src.quantum.qsed import QSEDRunner
from src.utils.visualization import plot_qsed_pipeline_results, plot_eight_directional_gradients
from src.utils.metrics import calculate_mse
```

## 1. NEQR Quantum Image Encoding Demonstration (4x4 Image)

```python
# Generate 4x4 synthetic image
img_4x4 = generate_synthetic_image(size=4, pattern='square')
print("4x4 Input Image Pixel Array:\n", img_4x4)

# Build NEQR quantum circuit
neqr = NEQRImage(img_4x4, q_bits=8)
qc = neqr.build_circuit()
print(f"NEQR Quantum Circuit Qubits: {qc.num_qubits}")
print(f"NEQR Gate Count Breakdown: {qc.count_ops()}")
```

## 2. Execute Full QSED Pipeline on 64x64 Image

```python
# Generate 64x64 synthetic test pattern
img_64 = generate_synthetic_image(size=64, pattern='checkerboard')

# Run QSED Runner
runner = QSEDRunner(q_bits=8, mode='hybrid')
results = runner.run_pipeline(img_64)

# Display stage-by-stage plot
plot_qsed_pipeline_results(results)
```

## 3. Display Eight Directional Gradient Responses

```python
plot_eight_directional_gradients(results['directional_gradients'])
```
